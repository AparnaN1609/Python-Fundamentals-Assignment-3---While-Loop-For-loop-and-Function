#### 1. Number Guessing Game

#### Create a number guessing game using a while loop and control statements
#### The program generates a random number between 1 and 10
#### The user gets three attempts to guess the correct number

```python
import random
```

```python
secret_number = random.randint(1, 10)
attempts = 3
```

```python
print("Welcome to the Number Guessing Game!")
print("Guess a number between 1 and 10.")
print("You have 3 attempts.")
```

```python
while attempts > 0:
    guess = int(input("Guess the number (between 1 and 10): "))

    if guess < 1 or guess > 10:
        print("Your guess is out of range. Please guess a number between 1 and 10.")
        continue

    attempts -= 1

    if guess == secret_number:
        print("Congratulations! You guessed the correct number.")
        break

    elif guess < secret_number:
        print("Too low. Try again.")

    else:
        print("Too high. Try again.")

else:
    print("Better luck next time! The number was", secret_number)
```

#### 2. Multiplication Table Generator

#### Generate a multiplication table from 1 to 10 for a given number using a for loop and the range() function

```python
num = int(input("Enter the number for which you want the multiplication table: "))

for i in range(1, 11):
    result = num * i
    print(f"{num} x {i} = {result}")
```

#### 3. BMI Calculator

#### Calculate Body Mass Index (BMI) using a function

```python
def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi
```

```python
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

bmi = calculate_bmi(weight, height)

print(f"Your BMI is: {bmi:.2f}")
```
